```python
# Function to add two matrices
def add_matrices(A, B):
    rows = len(A)
    cols = len(A[0])

    result = []

    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(A[i][j] + B[i][j])
        result.append(row)

    return result


# Main program
A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]

C = add_matrices(A, B)

print("First Matrix:")
for row in A:
    print(row)

print("Second Matrix:")
for row in B:
    print(row)

print("Addition of Two Matrices:")
for row in C:
    print(row)
```
