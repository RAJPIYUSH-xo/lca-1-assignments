```python
import re

# Function to validate PAN number
def validate_pan(pan):
    pattern = r"^[A-Z]{5}[0-9]{4}[A-Z]$"

    if re.fullmatch(pattern, pan):
        return True
    else:
        return False


# Main program
pan = input("Enter PAN Number: ")

if validate_pan(pan):
    print("Valid PAN Number")
else:
    print("Invalid PAN Number")
```
