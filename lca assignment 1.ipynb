```python
# Student Data Management using Dictionary, Tuple and List

# Function to add a student
def add_student(students, roll_no, name, branch, marks):
    student_details = (name, branch, marks)   # Tuple
    students[roll_no] = student_details


# Function to delete a student
def delete_student(students, roll_no):
    if roll_no in students:
        del students[roll_no]
    else:
        print("Student not found")


# Function to update student details
def update_student(students, roll_no, name, branch, marks):
    if roll_no in students:
        students[roll_no] = (name, branch, marks)
    else:
        print("Student not found")


# Function to display students
def display_students(students):
    for roll_no, details in students.items():
        print("Roll Number:", roll_no)
        print("Name:", details[0])
        print("Branch:", details[1])
        print("Marks:", details[2])
        print("--------------------")


# Main program

students = {}          # Dictionary

# Adding initial student records
add_student(students, 101, "Rahul", "CSE", 85)
add_student(students, 102, "Aman", "ENTC", 78)
add_student(students, 103, "Riya", "IT", 92)

# List of roll numbers
roll_numbers = [101, 102, 103]

# Add a new student
add_student(students, 104, "Sneha", "CSE", 88)
roll_numbers.append(104)

# Delete an existing student
delete_student(students, 102)
roll_numbers.remove(102)

# Update student details
update_student(students, 103, "Riya", "IT", 95)

# Display final records
print("Final Student Records:")
display_students(students)
```
